```python
# Descriptive statistics for numerical columns
numeric_cols = ["Quantity", "UnitPrice"]

df[numeric_cols].describe()
```

In above statistics we can see that in the Quantity column we have negative values present. That shouldn't be the case as that column description says "the quantities of each product (item) per transaction".
Same is the case with Unit Price column. Need to identify this thing and tackle it smartly.

```python
# Mode values for categorical columns
categorical_cols = ["StockCode", "CustomerID", "Country"]

df[categorical_cols].mode()
```

From above statistics we can understand which values appeared maximum time in the data. Through this we know that which StockCode is getting sold the most, which customer is ordering the maximum stock and which country is buying the most stock.

```python
# Negative Quantity (likely returns)
neg_quantity_df = df[df["Quantity"] < 0]

# Negative UnitPrice (less common, might be data issues)
neg_price_df = df[df["UnitPrice"] < 0]
```
