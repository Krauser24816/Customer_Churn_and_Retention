```python
# Calculate percentage of null values in specific columns

null_percentage = (
    df[["CustomerID", "Description"]]
    .isnull()
    .mean() * 100
)

# print(null_percentage)

# Create a summary table for null values

null_summary = pd.DataFrame({
    "Null Count": df[["CustomerID", "Description"]].isnull().sum(),
    "Null Percentage": (df[["CustomerID", "Description"]].isnull().mean() * 100)
})

null_summary
```

```python
# Rows where CustomerID is null

customerid_null_df = df[df["CustomerID"].isnull()]
```

```python
display(customerid_null_df)
```

```python
# Rows where both CustomerID and Description are null

null_customer_description_df = df[
    (df["CustomerID"].isnull()) & 
    (df["Description"].isnull())
]

display(null_customer_description_df)
```

```python
# Remove rows where BOTH CustomerID and Description are null

df = df[
    ~(
        (df["CustomerID"].isnull()) & 
        (df["Description"].isnull())
    )
]
```

```python
df.isnull().sum()
```

```python
# Find rows where CustomerID is null
null_customer_df = df[df["CustomerID"].isnull()].copy()

# Columns to compare (excluding CustomerID)
compare_cols = [col for col in df.columns if col != "CustomerID"]

# Find rows where CustomerID exists
non_null_customer_df = df[df["CustomerID"].notnull()].copy()

# Merge to identify matching rows
duplicate_like_nulls = null_customer_df.merge(
    non_null_customer_df,
    on=compare_cols,
    how="inner"
)

# Display matching rows
duplicate_like_nulls.head()
```

```python
duplicate_like_nulls.shape[0]
```

```python
# Rows where CustomerID is null
# AND Quantity = 0 OR UnitPrice = 0

null_customer_zero_df = df[
    (df["CustomerID"].isnull()) &
    (
        (df["Quantity"] <= 0) |
        (df["UnitPrice"] <= 0)
    )
]

null_customer_zero_df
```
